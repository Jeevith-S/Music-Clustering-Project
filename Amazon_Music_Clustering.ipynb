```python
features = [
    'danceability',
    'energy',
    'loudness',
    'speechiness',
    'acousticness',
    'instrumentalness',
    'liveness',
    'valence',
    'tempo',
    'duration_ms'
]
```

```python
X = one[features]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

#pca

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

one['PC1'] = X_pca[:, 0]
one['PC2'] = X_pca[:, 1]

print(one[['PC1', 'PC2']].head())
```

#elbow method

```python
from sklearn.cluster import KMeans

inertia_values = []

k_values = range(1, 11)

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42
    )

    model.fit(X_scaled)

    inertia_values.append(
        model.inertia_
    )
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.plot(
    k_values,
    inertia_values,
    marker="o"
)

plt.title("Elbow Method")

plt.xlabel("Number of Clusters (K)")

plt.ylabel("Inertia (SSE)")

plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score
)

silhouette_scores = []

for k in range(2,11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42
    )

    labels = kmeans.fit_predict(
        X_scaled
    )

    sil = silhouette_score(
        X_scaled,
        labels
    )

    db = davies_bouldin_score(
        X_scaled,
        labels
    )

    silhouette_scores.append(sil)   # ADD THIS

    print(f"K={k}")

    print(f"Silhouette = {sil:.4f}")

    print(f"Davies = {db:.4f}")

    print(f"Inertia = {kmeans.inertia_:.4f}")

    print("-"*30)
```

```python
import matplotlib.pyplot as plt

k_values = range(2,11)

plt.figure(figsize=(8,5))

plt.plot(
    k_values,
    silhouette_scores,
    marker="o"
)

plt.title(
    "Silhouette Score vs Number of Clusters"
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Silhouette Score"
)

plt.grid(True)

plt.show()
```

```python
# means of clusters
cluster_mean = one.groupby('cluster')[features].mean()

print(cluster_mean)
```

```python
#cluster names
def interpret_cluster(row):

    if row['speechiness'] >= 0.5:
        return "Speech-Heavy Tracks"

    elif row['energy'] >= 0.6 and row['danceability'] >= 0.6:
        return "Party / Upbeat Tracks"

    elif row['acousticness'] >= 0.6 and row['energy'] <= 0.4:
        return "Chill Acoustic Tracks"

    else:
        return "Mixed / General Tracks"


cluster_mean["cluster_description"] = (
    cluster_mean.apply(
        interpret_cluster,
        axis=1
    )
)

cluster_mean
```

#DBSCAN

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps=1.5,
    min_samples=5
)

dbscan_labels = dbscan.fit_predict(X_scaled)
```

```python
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score
)

def evaluate_dbscan(X_scaled, labels):

    sil_score = silhouette_score(
        X_scaled,
        labels
    )

    db_score = davies_bouldin_score(
        X_scaled,
        labels
    )

    print("DBSCAN Results")
    print("Silhouette Score :", sil_score)
    print("Davies-Bouldin   :", db_score)

    return sil_score, db_score
```
